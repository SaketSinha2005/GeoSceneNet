```python
from data import load_dataset
from model import build_mlp
```

```python
DATA_DIR = "D:\\GeoSceneNet\\data\\seg_train"   

train_df, val_df, class_names = load_dataset(
    data_dir=DATA_DIR,
    img_size=(150, 150),
    batch_size=32
)
```

```python
model, normalizor = build_mlp(learning_rate=1e-3, input_shape=(150, 150, 3), num_classes=6)
train_X = train_df.map(lambda x,y:x)
```

```python
normalizor.adapt(train_X)
history = model.fit(
    train_df, 
    validation_data=val_df,
    epochs=15
)
```

```python
MODEL_PATH = "../models/mlp_normalized.keras"

model.save(MODEL_PATH)
```
